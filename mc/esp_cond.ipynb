## I/ Espérance conditionnelle par rapport à une variable aléatoire

Une variable aléatoire X est une fonction qui à un évènement oméga associe sa probabilité (qui est une mesure sur l’espace des évènements).
Son espérance correspond à la valeur moyenne qu’elle prend, pondérée par sa probabilité :


$$
E[X] = \frac{\sum_{k} P(X = k)* k}{\sum_{i} P(X = i)} = \sum_{k} P(X = k)* k
$$

Intuitivement, l’espérance correspond à la valeur moyenne que prend la variable aléatoire sur un grand nombre de tirages (c’est ce que dit la loi des grands nombres, et plus largement le point de vue fréquentiste en probabilité).

---

La **probabilité conditionnelle** d’une variable aléatoire ($$ p(Y \mid X = 3)$$ ) correspond à la probabilité que Y prenne une certaine valeur sachant que X a une certaine valeur.

Par exemple, si Y=1 quand il pleut et Y=0 quand il faut beau, et si X donne la température, la probabilité qu’il pleuve sachant qu’il fait 3°C est ( $$p(Y=1 \mid X=3)$$ ), et cela définit plus largement une loi pour Y que l’on note ( $$P(Y \mid X=3)$$ ).

On peut modéliser l’information que X apporte d’une manière générale sur Y avec la variable ( $$P(Y \mid X) $$), qui donne la probabilité qu’il pleuve une fois qu’on connaît X.
Mais en général, on s’intéresse plutôt à ( $$P(Y = y_i \mid X) $$) qui est une fonction de X (si X=10°C, la probabilité qu’il pleuve sera plus élevée que si X=35°C).

---

L’espérance conditionnelle de $$( Y \mid X ) $$ notée $$ ( E[Y \mid X] )$$ modélise de même la valeur moyenne attendue pour Y sachant qu’on connaît la valeur de X.

Par exemple, si on achète des framboises, qui viennent de trois villes différentes où l'ensoleillement est différent :  une à 1km, une à 2km et une à 3km. À cause des variations d'ensoleillement, la teneur en sucre des framboises varie : elle vaut en moyenne 1g par framboise dans la ville 1, 5g dans la ville 2 et 10g dans la ville 3.

X modélise ici la distance de la ville où on a récolté des framboises.

Imaginons qu’on achète un panier de framboise, qui contient un mélange de framboises qui viennent des trois villes. On note Y la teneur en sucre d’une framboise de ce panier.

Alors, l’espérance conditionnelle de Y sachant X correspond à la teneur moyenne en sucre d’une framboise prise au hasard dans le panier.

On aura donc :

$$
E[Y \mid X] = E[Y \mid X=1] * p(X = 1) + E[Y \mid X=2] * p(X = 2) + E[Y \mid X=3] * p(X = 3)
$$

c’est à dire : la teneur moyenne en sucre des framboises du panier vaut la teneur moyenne en sucre des framboises dans qui viennent de la ville 1 * la probabilité d’avoir tirée une framboise cueillie dans la ville 1 + …

---

Si on s’intéresse à la variance de la quantité de sucre dans les framboises, ça va être la somme de deux termes : la variance de la teneur en sucre entre les différentes villes, et la variance de la teneur en sucre au sein d’une même ville (toutes les framboises de la ville 1 n’ont pas exactement la même teneur en sucre, la valeur 5g était une moyenne).

De là, on peut trouver intuitivement deux théorèmes : le théorème d’Adam et le théorème d’Eve.

---

### Loi d’Adam

$$
E[X] = E[E[X \mid Y]] = E[X \mid A_i] * P(A_i)
$$

---

### Loi d’Eve

La même mais pour les variances. Elle dit que la variance totale est la somme de la variance intra et inter variable.

$$
\text{var}(Y) = E(\text{var}(Y \mid X)) + \text{var}(E[Y \mid X])
$$

où le premier terme est la variance au sein d’une ville et le deuxième terme la variance d’une ville à l’autre.

---


## II/ L’espérance comme une projection

Si une variable aléatoire est intégrable deux fois (c’est-à-dire de variance finie), elle vit dans un espace de Hilbert $$( L^2(\Omega, \mathcal{F}, P) )$$.

Un élément d’un espace de Hilbert est un vecteur, et on peut définir sur cet espace une structure de produit scalaire qui permet de combiner ces vecteurs en les projetant les uns sur les autres.

Dans le cas des variables aléatoires, le produit scalaire est l’espérance :

$$
\langle X ; Y \rangle = E[XY]
$$

Par exemple, ( E[X] ) est le produit scalaire de X avec la variable aléatoire constante égale à 1, ce qui revient à projeter X sur l’espace des variables aléatoires constantes : l’espérance d’une variable aléatoire est sa partie constante.

$$
E[X] = E[1 * X]
$$

---

Si on centre une variable aléatoire (= on retire sa partie constante, donc son espérance), sa variance vaut ($$ E[X*X] $$) qui est donc la norme au carré de X.

La covariance  $$E[XY] - E[X]E[Y]$$  de deux variables centrées vaut $$ E[XY] $$ : c’est le produit scalaire de la partie non constante (= non prévisible) de la variable.

---

Ensuite, le coefficient de corrélation de Pearson

$$
\rho = \frac{\text{cov}(X,Y)}{\sqrt{\text{var}(X)} * \sqrt{\text{var}(Y)}}
$$

est le cosinus de l’angle formé par les variables X et Y ; et c’est pour ça qu’il est compris entre -1 et 1.

En effet,

$$
\text{cov}(X,Y) = ||X|| * ||Y|| * \cos(X,Y)
$$

où

$$
||.|| = \sqrt{\text{var}(.)}
$$

donc

$$
\rho = \cos(X,Y)
$$

---

Avoir une vision géométrique du problème permet souvent de s’en sortir plus facilement parce qu’on a plus d’intuition sur ce qui se passe.

# III/ Espérance conditionnelle par rapport à une tribu

## III/ Espérance conditionnelle par rapport à une tribu

Une tribu est une partie de l’ensemble des issues possibles d’une expérience (Omega) qui contient toute l’information sur laquelle l’expérience va permettre de trancher.

Par exemple, si on lance un dé mais qu’on ne peut connaître que la parité du résultat, on ne pourra trancher que sur la question : est-ce que le résultat est dans {2, 4, 6} ou est-ce qu’il est dans {1, 3, 5} ; mais on ne pourra pas dire si c’est un 2 ou un 4.

La tribu sera donc

$$
(\emptyset, {2,4,6}, {1,3,5}, \Omega)
$$

mais elle ne contiendra pas les singletons {2,4}.

Si une expérience permet de dire si le résultat est pair, elle permet nécessairement de dire s’il est impair, et c’est pour ça qu’on construit une tribu en réunissant toujours un évènement et son complémentaire.

La notion de tribu est donc reliée à celle d’information, et de ce qu’on peut distinguer.

---

Pour modéliser l’information à laquelle on a accès grâce à une expérience particulière, on peut projeter la variable aléatoire de toutes les issues possibles sur une tribu particulière.

D’un point de vue technique, la tribu est un sous-espace vectoriel de l’espace de Hilbert où vit la variable aléatoire, qui est tel que la projection de la variable aléatoire dans cet espace G est une variable aléatoire G-mesurable (ie ∀B∈B(R), X⁻¹(B) ∈ G), ce qui veut dire qu’elle peut être exprimée uniquement comme une fonction des éléments de G.

La projection est faite avec une espérance, comme on l’a vu plus haut.

En d’autres termes, on cherche à approximer la variable X en utilisant uniquement l'information donnée par G.

---

On a ainsi, pour toute variable aléatoire X et tribu G :

$$
X = E[X \mid G] + (X - E[X \mid G])
$$

où le premier terme est la projection de X sur G et le deuxième terme est la partie de X qui est orthogonale à G.

---

On obtient donc le théorème de Pythagore :

$$
E[X]^2 = E[X \mid G]^2 + (X - E[X \mid G])^2
$$

ce qui se réécrit :

$$
\text{var}(X) = \text{var}(E[X \mid G]) + E[\text{var}(X \mid G)]
$$

(on retrouve la loi d’Eve).

---

## Exemple

Pour un dé, pour lequel on n’a accès qu’à la parité du résultat :

$$
E[X \mid \text{pair}] = 4 = \frac{1}{3}(2+4+6)
$$

$$
E[X \mid \text{impair}] = 3
$$

Donc ( $$E[X \mid G]$$ ) est une variable aléatoire qui vaut 4 si X est pair et 3 si X est impair.

Son espérance est :

$$
E[E[X \mid G]] = p(\text{pair}) * 4 + p(\text{impair}) * 3 = 0.5 * 4 + 0.5 * 3 = 3.5
$$

Cela vaut exactement l’espérance de X.

---

Par ailleurs :

$$
E[E[X \mid G]] = E[X]
$$
 c’est la loi d’Adam avec des tribus.

---

$$ E[X \mid G] $$ est juste une variable aléatoire qui a le même comportement que X mais qui vit dans le sous-espace G.
